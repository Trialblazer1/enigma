Please run the below cell to see the Enigma visual simulator that opens up on a separate window.

The rotors are set to Left- I, Middle- II and Right- III and the Ring settings are at default of (1,1,1).

The initial settings of the rotors (from left to right) are set to 'AAZ".

The reflector used is B, and the plugleads are set to ('AB", 'CD', 'EF').

Please note that this simulation works only for 3 rotors at any given time.

The rotors that can be used for this simulation are any 3 in any order from the list of (I, II, III, IV, V).

This simulaton does not work on any other non-standard rotor or reflector components.

Each component is labelled according to it's functionality.

This simulation shows the path of the encryption- forward pass with green color, reflection with yellow color and backward pass with red color lines.

The characters being typed appear on the top of the window, with it's equivalent encrypted character right below the actual character being typed to be encrypted.

The simulation also works for sentences or words with spaces in between them. Example: how are you doing today.
